part2

```python
# CELL 5 — 生成 blob 测试数据
from sklearn.datasets import make_blobs
blob_centers = np.array(
    [[ 0.2,  2.3],
     [-1.5,  2.3],
     [-2.8,  1.8],
     [-2.8,  2.8],
     [-2.8,  1.3]])
blob_std = np.array([0.4, 0.3, 0.1, 0.1, 0.1])
X, y = make_blobs(n_samples=2000, centers=blob_centers,
                  cluster_std=blob_std, random_state=7)
print('Data shape:', X.shape)
```

```python
# CELL 6 — 可视化原始数据
def plot_clusters(X, y=None):
    plt.scatter(X[:, 0], X[:, 1], c=y, s=1)
    plt.xlabel('$x_1$', fontsize=14)
    plt.ylabel('$x_2$', fontsize=14, rotation=0)
plt.figure(figsize=(8, 4))
plot_clusters(X)
save_fig('blobs_plot')
plt.show()
```

```python
# CELL 7 — 训练 K-Means（核心代码！）
from sklearn.cluster import KMeans
k = 5
kmeans = KMeans(n_clusters=k, random_state=42)
y_pred = kmeans.fit_predict(X)
print('First 10 labels:', y_pred[:10])
print('Unique labels:', np.unique(y_pred))
```

```python
# CELL 8 — 查看质心坐标
print('Cluster centers found by K-Means:')
print(kmeans.cluster_centers_)
print('\nTrue blob centers (reference):')
print(blob_centers)
```

```python
# CELL 9 — 对新点预测所属簇
X_new = np.array([[0, 2], [3, 2], [-3, 3], [-3, 2.5]])
print('Predicted clusters for new points:', kmeans.predict(X_new))
```

```python
# CELL 10 — 定义画图辅助函数
def plot_data(X):
    plt.plot(X[:, 0], X[:, 1], 'k.', markersize=2)

def plot_centroids(centroids, weights=None, circle_color='w', cross_color='k'):
    if weights is not None:
        centroids = centroids[weights > weights.max() / 10]
    plt.scatter(centroids[:, 0], centroids[:, 1],
                marker='o', s=35, linewidths=8,
                color=circle_color, zorder=10, alpha=0.9)
    plt.scatter(centroids[:, 0], centroids[:, 1],
                marker='x', s=2, linewidths=12,
                color=cross_color, zorder=11, alpha=1)

def plot_decision_boundaries(clusterer, X, resolution=1000, show_centroids=True,
                             show_xlabels=True, show_ylabels=True):
    mins = X.min(axis=0) - 0.1
    maxs = X.max(axis=0) + 0.1
    xx, yy = np.meshgrid(np.linspace(mins[0], maxs[0], resolution),
                         np.linspace(mins[1], maxs[1], resolution))
    Z = clusterer.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    plt.contourf(Z, extent=(mins[0], maxs[0], mins[1], maxs[1]), cmap='Pastel2')
    plt.contour(Z, extent=(mins[0], maxs[0], mins[1], maxs[1]),
                linewidths=1, colors='k')
    plot_data(X)
    if show_centroids:
        plot_centroids(clusterer.cluster_centers_)
    if show_xlabels:
        plt.xlabel('$x_1$', fontsize=14)
    else:
        plt.tick_params(labelbottom=False)
    if show_ylabels:
        plt.ylabel('$x_2$', fontsize=14, rotation=0)
    else:
        plt.tick_params(labelleft=False)

print('Helper functions defined OK')
```

```python
# CELL 11 — 画 Voronoi 决策边界图
# 彩色区域=各簇地盘，圆圈叉=质心，黑线=簇边界
plt.figure(figsize=(8, 4))
plot_decision_boundaries(kmeans, X)
save_fig('voronoi_plot')
plt.show()
```

```python
# CELL 12 — Hard vs Soft Clustering
# Hard: 每个点只属于一个簇
# Soft: 每个点到所有质心的距离（保留更多信息）
distances = kmeans.transform(X_new)
print('Distance matrix shape:', distances.shape)
print('Distances to each centroid:')
print(distances.round(3))
print('Closest centroid for each point:', distances.argmin(axis=1))
```

```python
# CELL 13 — 验证 transform 就是欧氏距离
manual = np.linalg.norm(
    np.tile(X_new, (1, k)).reshape(-1, k, 2) - kmeans.cluster_centers_, axis=2)
print('Manual Euclidean distance:')
print(manual.round(3))
```

```python
# CELL 14 — 观察算法收敛过程
# max_iter=1/2/3 强行让算法只跑 N 轮，像慢动作一样逐步观察
kmeans_iter1 = KMeans(n_clusters=5, init='random', n_init=1,
                      algorithm='lloyd', max_iter=1, random_state=0)
kmeans_iter2 = KMeans(n_clusters=5, init='random', n_init=1,
                      algorithm='lloyd', max_iter=2, random_state=0)
kmeans_iter3 = KMeans(n_clusters=5, init='random', n_init=1,
                      algorithm='lloyd', max_iter=3, random_state=0)
kmeans_iter1.fit(X)
kmeans_iter2.fit(X)
kmeans_iter3.fit(X)
print('Done fitting 3 iterative models')
```

```python
# CELL 15 — 画 6 格对比图（收敛全过程）
# 行1: 随机初始化 | 第1次分配
# 行2: 第1次更新  | 第2次分配
# 行3: 第2次更新  | 第3次分配（基本稳定）
plt.figure(figsize=(10, 8))
plt.subplot(321)
plot_data(X)
plot_centroids(kmeans_iter1.cluster_centers_, circle_color='r', cross_color='w')
plt.ylabel('$x_2$', fontsize=14, rotation=0)
plt.tick_params(labelbottom=False)
plt.title('Update the centroids (initially randomly)', fontsize=14)
plt.subplot(322)
plot_decision_boundaries(kmeans_iter1, X, show_xlabels=False, show_ylabels=False)
plt.title('Label the instances', fontsize=14)
plt.subplot(323)
plot_decision_boundaries(kmeans_iter1, X, show_centroids=False, show_xlabels=False)
plot_centroids(kmeans_iter2.cluster_centers_)
plt.subplot(324)
plot_decision_boundaries(kmeans_iter2, X, show_xlabels=False, show_ylabels=False)
plt.subplot(325)
plot_decision_boundaries(kmeans_iter2, X, show_centroids=False)
plot_centroids(kmeans_iter3.cluster_centers_)
plt.subplot(326)
plot_decision_boundaries(kmeans_iter3, X, show_ylabels=False)
save_fig('kmeans_algorithm_plot')
plt.show()
```

part3

```python
def plot_clusterer_comparison(c1, c2, X, title1=None, title2=None):
    c1.fit(X)
    c2.fit(X)
    plt.figure(figsize=(10, 3.2))
    plt.subplot(121)
    plot_decision_boundaries(c1, X)
    if title1: plt.title(title1, fontsize=14)
    plt.subplot(122)
    plot_decision_boundaries(c2, X, show_ylabels=False)
    if title2: plt.title(title2, fontsize=14)

kmeans_rnd_init1 = KMeans(n_clusters=5, init='random', n_init=1,
                          algorithm='lloyd', random_state=2)
kmeans_rnd_init2 = KMeans(n_clusters=5, init='random', n_init=1,
                          algorithm='lloyd', random_state=5)
plot_clusterer_comparison(kmeans_rnd_init1, kmeans_rnd_init2, X,
                          'Solution 1', 'Solution 2 (different random init)')
save_fig('kmeans_variability_plot')
plt.show()
```

```python
# CELL 17 — Inertia 比较
print('Good K-Means inertia:   ', kmeans.inertia_)
print('Random init 1 inertia:  ', kmeans_rnd_init1.inertia_)
print('Random init 2 inertia:  ', kmeans_rnd_init2.inertia_)
print('Lower inertia = tighter clusters = better')
```

```python
# CELL 18 — 解决方案：n_init=10 多次初始化取最优
kmeans_rnd_10_inits = KMeans(n_clusters=5, init='random', n_init=10,
                              algorithm='lloyd', random_state=2)
kmeans_rnd_10_inits.fit(X)
print('n_init=10 inertia:', kmeans_rnd_10_inits.inertia_)
plt.figure(figsize=(8, 4))
plot_decision_boundaries(kmeans_rnd_10_inits, X)
plt.title('n_init=10, result is more stable')
plt.show()
```

```python
# CELL 19 — K-Means++ 默认参数查看
print(KMeans())  # 默认 init='k-means++', n_init=10
good_init = np.array([[-3, 3], [-3, 2], [-3, 1], [-1, 2], [0, 2]])
kmeans_good = KMeans(n_clusters=5, init=good_init, n_init=1, random_state=42)
kmeans_good.fit(X)
print('Good init inertia:', kmeans_good.inertia_)
```

```python
# CELL 22 — 演示 k=3 和 k=8 都不对
kmeans_k3 = KMeans(n_clusters=3, random_state=42)
kmeans_k8 = KMeans(n_clusters=8, random_state=42)
plot_clusterer_comparison(kmeans_k3, kmeans_k8, X, '$k=3$', '$k=8$')
save_fig('bad_n_clusters_plot')
plt.show()
print('k=3 inertia:', kmeans_k3.inertia_)
print('k=8 inertia:', kmeans_k8.inertia_)
```

```python
# CELL 23 — Elbow Method（肘部法则）
kmeans_per_k = [KMeans(n_clusters=k, random_state=42).fit(X)
                for k in range(1, 10)]
inertias = [model.inertia_ for model in kmeans_per_k]
plt.figure(figsize=(8, 3.5))
plt.plot(range(1, 10), inertias, 'bo-')
plt.xlabel('$k$', fontsize=14)
plt.ylabel('Inertia', fontsize=14)
plt.annotate('Elbow',
             xy=(4, inertias[3]),
             xytext=(0.55, 0.55),
             textcoords='figure fraction',
             fontsize=16,
             arrowprops=dict(facecolor='black', shrink=0.1))
plt.axis([1, 8.5, 0, 1300])
save_fig('inertia_vs_k_plot')
plt.show()
```

```python
# CELL 24 — Silhouette Score 概念
# 接近+1: 点在自己簇内很紧密，离其他簇很远
# 接近 0: 点在簇边界上
# 接近-1: 点可能被分到了错误的簇
from sklearn.metrics import silhouette_score
score = silhouette_score(X, kmeans.labels_)
print('k=5 Silhouette Score:', round(score, 4))
print('(closer to 1 = better)')
```

```python
# CELL 25 — 画 Silhouette Score vs k 曲线
silhouette_scores = [silhouette_score(X, model.labels_)
                     for model in kmeans_per_k[1:]]
plt.figure(figsize=(8, 3))
plt.plot(range(2, 10), silhouette_scores, 'bo-')
plt.xlabel('$k$', fontsize=14)
plt.ylabel('Silhouette score', fontsize=14)
plt.axis([1.8, 8.5, 0.55, 0.7])
save_fig('silhouette_score_vs_k_plot')
plt.show()
```

```python
# CELL 26 — Silhouette 分析图（k=3,4,5,6 对比）
from sklearn.metrics import silhouette_samples
from matplotlib.ticker import FixedLocator, FixedFormatter
plt.figure(figsize=(11, 9))
for k in (3, 4, 5, 6):
    plt.subplot(2, 2, k - 2)
    y_pred = kmeans_per_k[k - 1].labels_
    silhouette_coefficients = silhouette_samples(X, y_pred)
    padding = len(X) // 30
    pos = padding
    ticks = []
    for i in range(k):
        coeffs = silhouette_coefficients[y_pred == i]
        coeffs.sort()
        color = mpl.cm.Spectral(i / k)
        plt.fill_betweenx(np.arange(pos, pos + len(coeffs)), 0, coeffs,
                          facecolor=color, edgecolor=color, alpha=0.7)
        ticks.append(pos + len(coeffs) // 2)
        pos += len(coeffs) + padding
    plt.gca().yaxis.set_major_locator(FixedLocator(ticks))
    plt.gca().yaxis.set_major_formatter(FixedFormatter(range(k)))
    if k in (3, 5): plt.ylabel('Cluster')
    if k in (5, 6):
        plt.gca().set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])
        plt.xlabel('Silhouette Coefficient')
    else:
        plt.tick_params(labelbottom=False)
    plt.axvline(x=silhouette_scores[k - 2], color='red', linestyle='--')
    plt.title('$k={}$'.format(k), fontsize=16)
save_fig('silhouette_analysis_plot')
plt.show()
```

```python
# CELL 30 — 加载图片并查看形状
from matplotlib.image import imread
image = imread(os.path.join(images_path, filename))
print('Image shape:', image.shape)  # (H, W, 3)
plt.figure(figsize=(6, 4))
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
# CELL 31 — 图像分割核心代码
# 每个像素的 RGB 值作为一个数据点，聚成 k 类
X_img = image.reshape(-1, 3)
print('Pixel data shape:', X_img.shape)
kmeans_img = KMeans(n_clusters=8, random_state=42).fit(X_img)
segmented_img = kmeans_img.cluster_centers_[kmeans_img.labels_]
segmented_img = segmented_img.reshape(image.shape)
plt.figure(figsize=(8, 4))
plt.subplot(121)
plt.imshow(image)
plt.title('Original')
plt.axis('off')
plt.subplot(122)
seg_show = (segmented_img / 255.0) if segmented_img.max() > 1 else segmented_img
plt.imshow(seg_show)
plt.title('Segmented (8 colors)')
plt.axis('off')
plt.show()
```

```python
# CELL 32 — 对比不同颜色数量的分割效果
segmented_imgs = []
n_colors = (10, 8, 6, 4, 2)
for n_clusters in n_colors:
    km = KMeans(n_clusters=n_clusters, random_state=42).fit(X_img)
    seg = km.cluster_centers_[km.labels_]
    segmented_imgs.append(seg.reshape(image.shape))
plt.figure(figsize=(10, 5))
plt.subplots_adjust(wspace=0.05, hspace=0.1)
plt.subplot(231)
plt.imshow(image)
plt.title('Original image')
plt.axis('off')
for idx, n_clusters in enumerate(n_colors):
    plt.subplot(232 + idx)
    img_s = segmented_imgs[idx]
    plt.imshow((img_s / 255.0) if img_s.max() > 1 else img_s)
    plt.title('{} colors'.format(n_clusters))
    plt.axis('off')
save_fig('image_segmentation_diagram', tight_layout=False)
plt.show()
```

```python
# CELL 33 — 加载手写数字数据集并切分
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
X_digits, y_digits = load_digits(return_X_y=True)
print('Shape:', X_digits.shape)  # 1797 images, 64 pixels each
X_train, X_test, y_train, y_test = train_test_split(X_digits, y_digits, random_state=42)
print('Train:', X_train.shape, 'Test:', X_test.shape)
```

```python
# CELL 34 — 基准模型：直接逻辑回归
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(solver='lbfgs', max_iter=5000, random_state=42)
log_reg.fit(X_train, y_train)
log_reg_score = log_reg.score(X_test, y_test)
print('Direct LogReg accuracy:', round(log_reg_score, 4))
```

```python
# CELL 35 — 改进：K-Means + 逻辑回归 Pipeline
#先单独用 Logistic Regression 分类 digits 数据集，记录准确率。
#再用 Pipeline 先 K-Means（k=50）再 Logistic Regression，对比准确率提升。最后用 GridSearchCV 找最优 k。
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('kmeans', KMeans(n_clusters=50, random_state=42)),
    ('log_reg', LogisticRegression(solver='lbfgs', max_iter=5000, random_state=42)),
])
pipeline.fit(X_train, y_train)
pipeline_score = pipeline.score(X_test, y_test)
print('K-Means + LogReg accuracy:', round(pipeline_score, 4))
improvement = 1 - (1 - pipeline_score) / (1 - log_reg_score)
print('Error rate reduced by:', round(improvement * 100, 1), '%')
```

```python
# CELL 36 — GridSearchCV 找最优 k
from sklearn.model_selection import GridSearchCV
param_grid = dict(kmeans__n_clusters=range(2, 100))
grid_clf = GridSearchCV(pipeline, param_grid, cv=3, verbose=2)
grid_clf.fit(X_train, y_train)
print('Best params:', grid_clf.best_params_)
print('Best accuracy:', round(grid_clf.score(X_test, y_test), 4))
```

```python
# CELL 37 — 生成月牙形数据
from sklearn.datasets import make_moons
X, y = make_moons(n_samples=1000, noise=0.05, random_state=42)
plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=y, s=10)
plt.title('Moon-shaped data (K-Means fails here)')
plt.show()
```

```python
# CELL 38 — 运行 DBSCAN（eps=0.05）
# eps: 邻域半径；min_samples: 核心点最少邻居数
# 标签 -1 = 噪声点
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps=0.05, min_samples=5)
dbscan.fit(X)
print('First 10 labels:', dbscan.labels_[:10])
print('Unique labels:', np.unique(dbscan.labels_))
print('Num core points:', len(dbscan.core_sample_indices_))
```

```python
# CELL 39 — 用更大的 eps=0.2
dbscan2 = DBSCAN(eps=0.2)
dbscan2.fit(X)
print('Labels with eps=0.2:', np.unique(dbscan2.labels_))
print('Num core points:', len(dbscan2.core_sample_indices_))
```

```python
# CELL 40 — 可视化对比 eps=0.05 vs eps=0.2
# 大圆圈=核心点，红叉=噪声点，小点=边界点
def plot_dbscan(dbscan, X, size, show_xlabels=True, show_ylabels=True):
    core_mask = np.zeros_like(dbscan.labels_, dtype=bool)
    core_mask[dbscan.core_sample_indices_] = True
    anomalies_mask = dbscan.labels_ == -1
    non_core_mask = ~(core_mask | anomalies_mask)
    cores = dbscan.components_
    anomalies = X[anomalies_mask]
    non_cores = X[non_core_mask]
    plt.scatter(cores[:, 0], cores[:, 1],
                c=dbscan.labels_[core_mask], marker='o', s=size, cmap='Paired')
    plt.scatter(cores[:, 0], cores[:, 1],
                marker='*', s=20, c=dbscan.labels_[core_mask])
    plt.scatter(anomalies[:, 0], anomalies[:, 1], c='r', marker='x', s=100)
    plt.scatter(non_cores[:, 0], non_cores[:, 1],
                c=dbscan.labels_[non_core_mask], marker='.')
    if show_xlabels: plt.xlabel('$x_1$', fontsize=14)
    else: plt.tick_params(labelbottom=False)
    if show_ylabels: plt.ylabel('$x_2$', fontsize=14, rotation=0)
    else: plt.tick_params(labelleft=False)
    plt.title('eps={:.2f}, min_samples={}'.format(dbscan.eps, dbscan.min_samples), fontsize=14)

plt.figure(figsize=(9, 3.2))
plt.subplot(121)
plot_dbscan(dbscan, X, size=100)
plt.subplot(122)
plot_dbscan(dbscan2, X, size=600, show_ylabels=False)
save_fig('dbscan_plot')
plt.show()
```

```python
# CELL 41 — DBSCAN + KNN：预测新点
# DBSCAN 没有 predict()，用 KNN 在核心点上训练来预测
from sklearn.neighbors import KNeighborsClassifier
dbscan = dbscan2
knn = KNeighborsClassifier(n_neighbors=50)
knn.fit(dbscan.components_,
        dbscan.labels_[dbscan.core_sample_indices_])
X_new = np.array([[-0.5, 0], [0, 0.5], [1, -0.1], [2, 1]])
print('Predicted clusters:', knn.predict(X_new))
print('Probabilities:')
print(knn.predict_proba(X_new).round(3))
```

```python
# CELL 42 — 画决策边界
plt.figure(figsize=(6, 3))
plot_decision_boundaries(knn, X, show_centroids=False)
plt.scatter(X_new[:, 0], X_new[:, 1], c='b', marker='+', s=200, zorder=10)
save_fig('cluster_classification_plot')
plt.show()
```

```python
# CELL 43 — 处理离群点：距离 > eps 标为噪声
y_dist, y_pred_idx = knn.kneighbors(X_new, n_neighbors=1)
y_pred = dbscan.labels_[dbscan.core_sample_indices_][y_pred_idx]
y_pred[y_dist > 0.2] = -1
print('Final predictions (with outlier detection):', y_pred.ravel())
print('(-1 = outlier / noise)')
```
